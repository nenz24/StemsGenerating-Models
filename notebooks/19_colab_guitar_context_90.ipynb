Base / LEGO / CUDA FP32 / seed 0 / CFG 7 / 50 langkah; caption gitar elektrik clean tetap sama. Jalankan sel ini lagi jika satu chunk gagal: chunk selesai dipakai ulang setelah verifikasi. ## Simpan hasil dan kirim feedback

Simpan `guitar_chunked_result.zip`: WAV gitar tiap chunk, WAV gitar/mix penuh, serta report. Kirim `guitar_chunked_feedback.zip`: MP3 gitar/mix penuh, preview chunk sebelum stitching, WAV sambungan dan report aktual. Paket ini tidak memuat bobot, piano asli, atau track bass/drum. Untuk mix gabungan nanti, simpan juga lossless result ZIP bass dan `drum_shared_kit_result.zip` yang sudah disetujui.


```python
stitched = GUITAR_CHUNK_OUT / "stitched"
if (stitched / "seams.json").is_file():
    for filename, full_result in [("guitar_chunked_feedback.zip", False), ("guitar_chunked_result.zip", True)]:
        bundle = GUITAR_CHUNK_OUT / filename
        with zipfile.ZipFile(bundle, "w", compression=zipfile.ZIP_DEFLATED) as archive:
            archive.write(GUITAR_CHUNK_PLAN, "plan.json")
            archive.write(GUITAR_CHUNK_OUT / "progress.json", "progress.json")
            for path in sorted(stitched.rglob("*")):
                if path.is_file() and (path.suffix in {".json", ".mp3"} or path.name.startswith("seam_") or full_result and path.suffix == ".wav"):
                    archive.write(path, "stitched/" + str(path.relative_to(stitched)))
            if full_result:
                saved = json.loads((GUITAR_CHUNK_OUT / "progress.json").read_text())
                for c in guitar_plan["chunks"]:
                    folder = (GUITAR_CHUNK_OUT / saved[str(c["index"])]).resolve()
                    assert folder.is_relative_to(GUITAR_CHUNK_OUT / f'chunk_{c["index"]:03d}')
                    stem = context.verify_chunk(folder, guitar_plan, c)
                    archive.write(stem, f'chunks/chunk_{c["index"]:03d}_guitar.wav')
        print(filename, round(bundle.stat().st_size/1024**2, 2), "MiB")
        files.download(str(bundle))
else:
    print("Belum ada hasil lengkap. Tidak ada audio hasil yang dibuat oleh dry-run.")
```
